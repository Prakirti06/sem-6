```python
import pandas as pd
import numpy as np

df = pd.read_csv("cardio_train.csv",sep = ';')

# Age in integer years
df['age'] = (df['age'] / 365).round().astype(int)

binary_cols = ['gender', 'smoke', 'alco', 'active', 'cardio']
num_cols = ['age', 'height', 'weight', 'ap_hi', 'ap_lo']

# Missing values
for col in num_cols:
    df[col] = df[col].fillna(df[col].median())

for col in binary_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

# Outlier handling
for col in ['height', 'weight', 'ap_hi', 'ap_lo']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    df[col] = np.clip(df[col], Q1 - 1.5*IQR, Q3 + 1.5*IQR)

# ✅ THIS dataset keeps real values
df_eda = df.copy()
```

```python
from sklearn.preprocessing import StandardScaler

df_ml = df.copy()

scale_cols = ['height', 'weight', 'ap_hi', 'ap_lo']

scaler = StandardScaler()
df_ml[scale_cols] = scaler.fit_transform(df_ml[scale_cols])
```
