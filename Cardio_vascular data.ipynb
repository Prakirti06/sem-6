## Step-11 Display histogram of choresterol distribution.

```python
plt.figure(figsize=(5,6))
plt.hist(data['cholesterol'])
plt.xlabel("cholesterol")
plt.ylabel("Frequency")
plt.title("Cholestrol Distribution")
plt.show()
```

## Step-12 Display scatter plot age vs cardio.

```python
plt.figure(figsize=(10,6))
plt.scatter(data['age'], data['cardio'])
plt.xlabel("Age")
plt.ylabel("Cardio")
plt.title("Age vs Cardio")
plt.show()
```

## Step-13 Display boxplot of cholesterol by cardio.

```python
plt.figure(figsize=(8,5))
data.boxplot(column='cholesterol', by='cardio')
plt.xlabel("Cardio (0 = No Disease, 1 = Disease)")
plt.ylabel("Cholestrol")
plt.title("cholesterol by Cardio Disease")
plt.show()
```
